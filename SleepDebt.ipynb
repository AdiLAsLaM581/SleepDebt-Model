```python
#checking duplicates, missing values and then drop useless column like userid as it doesn't affect the result of the model
#And the error showing below is because i run the cell twice to double check the output but the cell already deleted userId column so column doesn't exist error is there....
```

```python
#Duplicates
print('Duplicate rows found:', df.duplicated().sum())
df = df.drop_duplicates()

#Count missing values
missing_counts = df.isnull().sum()
missing_cols = missing_counts[missing_counts > 0]

plt.figure(figsize=(6, 4))
missing_cols.plot(kind='bar')
plt.title('Missing Values Count')
plt.xlabel('Columns')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()

#Drop useless column (ID column)
df = df.drop(columns=['user_id'])
print("Dropped 'user_id' column")
```

```python
#Filling Numeric values with median as its robust against skewed data and outliers
#Filling categorical missing values using mode (most frequent app category)
```

```python

df['sleep_latency_min'].fillna(df['sleep_latency_min'].median(), inplace=True)
df['caffeine_post_5pm_mg'].fillna( df['caffeine_post_5pm_mg'].median(), inplace=True)

df['primary_bedtime_app'].fillna(
    df['primary_bedtime_app'].mode()[0], inplace=True
)
print('Remaining missing values:', df.isnull().sum().sum())
```

```python
#creating feauture:
# F1==> Total Screen Impact (Phone minutes adjusted by screen brightness)
# F2==> Sleep Efficiency Ratio (Ratio of time slept versus total time in bed)
```

```python


df['screen_impact_score'] = df['bedtime_phone_minutes'] * (
    df['screen_brightness_pct'] / 100
)


df['sleep_efficiency_ratio'] = df['total_sleep_hours'] / (
    (df['sleep_latency_min'] / 60) + df['total_sleep_hours']
)

print('New features created successfully!')
display(
    df[[
        'bedtime_phone_minutes',
        'screen_brightness_pct',
        'screen_impact_score',
        'sleep_efficiency_ratio',
    ]].head()
)
```

```python
#Select X(feature) and Y(target)
y = df['sleep_debt_category']
X = df.drop(columns=['sleep_debt_category'])

print("Features shape (X):", X.shape)
print("Target shape (y):", y.shape)
```

```python
#Split data into 80%/20% train and test sets as per the given condition
```

```python
from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Print sizes
print(f'Train set size (X_train): {X_train.shape}')
print(f'Test set size (X_test): {X_test.shape}')

print('\nTrain Class Proportions (%):')
print(y_train.value_counts(normalize=True) * 100)
print('\nTest Class Proportions (%):')
print(y_test.value_counts(normalize=True) * 100)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

num_cols = X.select_dtypes(include=['int64', 'float64']).columns
cat_cols = X.select_dtypes(include=['object']).columns

#Pipeline for Numbers: Fill missing with median, then scale
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),
])

#Pipeline for Text/Categories: Fill missing with mode, then One-Hot Encode
cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False)),
])

#Combineboth pipelines using ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_pipeline, num_cols),
        ('cat', cat_pipeline, cat_cols),
    ]
)
```

```python
#Fit and transform on train data:
X_train_processed = preprocessor.fit_transform(X_train)

#Transform test data only
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing complete!")
print("X_train processed shape:", X_train_processed.shape)
print("X_test processed shape:", X_test_processed.shape)
```

```python
# Step 7: Final checks
#final shapes
print('--- 1. FINAL SHAPES ---')
print('X_train_processed shape:', X_train_processed.shape)
print('X_test_processed shape :', X_test_processed.shape)

#Confirm missing vales
print('\n--- 2. MISSING VALUES CHECK ---')
print('Missing values in X_train_processed:', np.isnan(X_train_processed).sum())
print('Missing values in X_test_processed :', np.isnan(X_test_processed).sum())

#Confirmation statement
print('\n--- 3. DATA LEAKAGE CONFIRMATION ---')
print(
    'Confirmed: Preprocessor was fit ONLY on X_train. X_test was transformed'
    ' using parameters learned from train data.'
)
```
